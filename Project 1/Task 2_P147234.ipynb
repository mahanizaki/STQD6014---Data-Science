# Part 2: List, Dictionaries & Looping

```python
products={}

prompt='\nProduct Price Management System:'
prompt+='\n1. Add a new product'
prompt+='\n2. Update a product\'s price or quantity'
prompt+='\n3. View all products'
prompt+='\n4. Stop the program'
prompt+='\nEnter your choice:'

active=True
while active:
    system=input(prompt).strip()

    if system=='1':
        product_name=input('Enter product name:').strip()
        try:
            price=float(input('Enter product price:').strip())
            quantity=int(input('Enter product quantity:').strip())
            products[product_name]={'price':price,'quantity':quantity}
            print(f'Product "{product_name}" added successfully.')
        except ValueError:
            print('Invalid input for price or quantity. Please try again.')

    elif system=='2':
        product_name=input('Enter product name to update:').strip()
        if product_name in products:
            try:
                updated_price=input(f'Enter updated price for "{product_name}" (or press Enter to skip):').strip()
                updated_quantity=int(input(f'Enter updated quantity for "{product_name}" (or press Enter to skip):').strip())

                if updated_price:
                    products[product_name]['price']=float(updated_price)
                if updated_quantity:
                    products[product_name]['quantity']=int(updated_quantity)
                    
                print(f'Product {product_name} updated successfully.')
            except ValueError:
                print('Invalid input for price or quantity. Please try again.')

        else:
            print(f'Product "{product_name}" not found.')

    elif system=='3':
        if not products:
            print('No products added yet.')
        else:
            print('View all products:')
            for product, details in products.items():
                print(f'{product}: RM{details["price"]:.2f},Quantity:{details["quantity"]}')
    
    elif system=='4':
        print('Exit the program.')
        break
    
    else:
        print('Invalid input for system. Please choose between 1-4.')
```
