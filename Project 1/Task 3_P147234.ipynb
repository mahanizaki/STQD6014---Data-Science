### Question 2:

```python
#Parent class
class Cat():
    def __init__(self,breed,name):
        self.breed=breed
        self.name=name
    def describe_cat(self):
        print("\nI have a "+self.breed.title()+" cat.")
        print("Its name is "+self.name.title()+".")

#Child class
class Kitten(Cat):
    def __init__(self,breed,name,age):
        super().__init__(breed,name)
        self.age=age
    def describe_kitten(self):
        print("Its age is "+str(self.age)+" years old.")
```
